```python
# ------- inputs -------------------------------------------
analyst_name = input("Enter name of the analyst: ")
region = input("Enter the region: ")0

#------- calculations----------------------------------------
def all_functions(sales):
    mode = mode(sales)
    median = median(sales)
    mean = mean(sales)
    highest = max(sales)
    lowest = min(sales)
    total = sum(sales)
    stdev = stdev(sales)
    return mean, median, mode, stdev, total, highest, lowest
    
#-------- unpack the function/ call function----- allows you to call a specific function
#----- to unpack you have to assign the specific variable to the function
all_functions(sales) = mean, median, mode, stdev, total, highest, lowest 

```

```python

def analyze_sales(analyst, region, sales): 
    mean   = statistics.mean(sales) 
    median = statistics.median(sales) 
    mode   = statistics.mode(sales) 
    stdev  = statistics.stdev(sales) 
    total  = sum(sales)                 # built-in — no import needed 
    high   = max(sales)                 # built-in 
    low    = min(sales)                 # built-in 
    return mean, median, mode, stdev, total, high, low 
 
# ── Collect input ────────────────────────────────────────────── 
analyst = input("Analyst name:  ") 
region  = input("Region:        ") 
 
print("Enter daily sales for 7 days (one per line):") 
sales = [float(input(f"  Day {i+1}: $")) for i in range(7)] 
 
# ── Call the function ────────────────────────────────────────── 
mean, median, mode, stdev, total, high, low = analyze_sales(analyst, 
region, sales) 
 
# ── Print formatted report ───────────────────────────────────── 
print(f""" 
======= Weekly Sales Statistics Report ======= 
Analyst : {analyst} 
Region  : {region} 
Data    : {sales} ----------------------------------------------- 
Total revenue : ${total:.2f} 
Mean (avg)    : ${mean:.2f} 
Median        : ${median:.2f} 
Mode          : ${mode:.2f} 
Std deviation : ${stdev:.2f} 
Highest day   : ${high:.2f} 
Lowest day    : ${low:.2f} """)
```
