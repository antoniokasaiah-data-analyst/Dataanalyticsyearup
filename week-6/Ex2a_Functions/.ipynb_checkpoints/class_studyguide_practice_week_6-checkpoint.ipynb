## Default Parameters

```python
def greet4(a= "kind",b="day"):
    return f"You are very {a} on this {b}"
# sets a default value for the parameters 
# allows you to call the function without having to add values the parameteres
# because the parameters have values already
```

```python
print(greet4())
print(greet4("a","b"))
# if you print without specifying the parameters both are printed 
```

```python
print(greet4("a"))
print(greet4("himynameis"))
# you can print only specifying one parameters
```

### Arbitrary Arguments

```python
def greet5(*args):
    return args
print(greet5('power','lift'))
# allows you to input any amount of values after function is created 
# and it is stored as a tuple
# parameters are a special type of variabe used in a function
```

```python
for name in greet5('power','supe'):
    print(f"hi, {name}")
h1 = greet5('sup','cake')
for name in h1:
    print(f'hi {name}')
# you can use the function directly with the assigned parameters in a for loop
# or you can assign the function to a variable and define the parameters
# and use that in a for loop
```

```python
def greet6(*args):
    return args
greet = greet6()
for name in greet:
    print(f'hi {name}')
# since there is no values in the parameters nothiing is printed
```

```python
def greet7(**kwargs):
    return kwargs
print(greet7(name = 'kaz',age = 10, race = 'black'))
# Allows you to create a dictionary 
# you can add the keys and values after the function is created 
# or the keys before its created 
def greet8(a,b,**kwargs):
    kwargs['name'] = a
    kwargs['age'] =b
    return kwargs
print(greet8())
# if you dont add the values to the keys you get an error 
# the keys are the values in the closed brackets next to kwargs 
```

```python
def greet8(a,b,**kwargs):
    kwargs['name'] = a
    kwargs['age'] =b
    return kwargs
print(greet8('kaz','6ft'))
```

```python
def greet8(a= 'sup',b= 'wow',**kwargs):
    kwargs['name'] = a
    kwargs['age'] =b
    return kwargs
print(greet8('kaz','6ft'))
print(greet8())
# default values still work with arbitrary arguments
```

```python
def greet9(name):
    for n in range(name):
        return(f" my {name}")
# range indexes integers not strings
# you have to use the len function
```

```python
print(greet9(name))
```

```python
def greet9(name):
    for n in range(len(name)):
        return(f" my {name}")
print(greet9('power'))
```

```python
for i in range(5):
    print(i)
```

```python
def greet9(name):
    for n in range(len(name)):
        print(f" my {name}")
print(greet9('power'))
```

```python
for i in range(len('antonio')):
    print(f' My name starts with {'antonio'}')
# without brackets loop iterates antonio 7 times because 
# antonio has 7 characters that are indexed
```

```python
for i in range(len('antonio')):
    print(f' My name starts with {'antonio'[i]}')
# iterates each character that was indexed 
```

```python
len('antonio')
```

```python
for t in range(len('kasaiah')):
    if t == 2:
        print(f' the second letter in my name is {'kasaiah'[t]}')
# variables that are strings need to be in a parenthesis 
# prints character at index 2 which is s
```

```python
for t in range(len('kasaiah')):
    if 'kasaiah'[t] == 2:
        print(f' the second letter in my name is {'kasaiah'[t]}')
# prints blank because 'kasaiah'[t] indexes characters in kasaiah 
# which is k a s a i a h
# these characters can never eqaul 2
# so code doesnt print value 
```

```python
for t in range(len('kasaiah')):
    if t >= 2:
        print(f' the second letter in my name is {'kasaiah'[t]}')
        print(t)
# prints values that are greater or equal to t
# t is the index number for the string kasaiah 
```
